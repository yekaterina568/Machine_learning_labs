```python
import math
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + math.exp(-z))

x1 = 1
x2 = 0
w1 = 0.6
w2 = 0.4
b = -0.2

z = x1 * w1 + x2 * w2 + b
prediction = sigmoid(z)
predicted_class = round(prediction)
print("z =", z)
print("prediction =", prediction)
print("predicted class =", predicted_class)
```

```python
y = 1
loss = (y - prediction)**2
print("Correct target =", y)
print("Prediction =", prediction)
print("Loss =", round(loss, 3))
```

```python
experiments = [
    {"w1": 0.6, "w2": 0.4, "b": -0.2},
    {"w1": 1.0, "w2": 0.4, "b": -0.2},
    {"w1": 0.6, "w2": 0.4, "b": -1.0},
    {"w1": 1.5, "w2": 0.4, "b": 0.2},
]

res = []

for e in experiments:
    z = x1*e["w1"] + x2*e["w2"] + e["b"]
    pred = sigmoid(z)
    loss  = (y-pred)**2

    res.append({
        "w1": e["w1"],
        "w2": e["w2"],
        "b": e["b"],
        "z": z,
        "pred": pred,
        "loss": loss
    })

res_df = pd.DataFrame(res)
display(res_df.round(4))
```

```python
def warm_membership(temp):
    if temp <= 10 or temp >= 40:
        return 0
    elif 10 < temp < 25:
        return (temp - 10)/(25-10)
    elif temp == 25:
        return 1
    else:
        return (40 - temp)/(40-25)

for temp in [10, 15, 20, 25, 30, 35, 40]:
    print(f"{temp}°C -> Warm = {round(warm_membership(temp), 3)}")
```

```python
temps = np.linspace(0, 50, 101)
warm_values = [warm_membership(t) for t in temps]
plt.plot(temps, warm_values)
plt.xlabel("Temperature (C)")
plt.ylabel("Warm membership")
plt.title('Triangle membership function: Warm')
plt.ylim(-0.05, 1.05)
plt.grid(True)
plt.show()
```

```python
hot = 0.8
humid = 0.6

fuzzy_and = min(hot, humid)
fuzzy_or = max(hot, humid)
not_hot = 1-hot
not_humid = 1-humid

print(f'Fuzzy hot and humid: {fuzzy_and}')
print(f'Fuzzy hot or humid: {fuzzy_or}')
print(f'Fuzzy not hot: {round(not_hot, 2)}')
print(f'Fuzzy not humid: {not_humid}')
```

```python
rule_strength = fuzzy_and
print(f'Rule strength = {rule_strength}')
```
